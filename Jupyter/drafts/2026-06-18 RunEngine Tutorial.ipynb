# Following the RunEngine Tutorial

I will be following Bluesky's official tutorial on the RunEngine.

## References
1. [Bluesky RunEngine Tutorial](https://blueskyproject.io/bluesky/main/tutorial.html#the-runengine)

```python
from bluesky import RunEngine

RE = RunEngine({})
```

## Best effort callbac

```python
from bluesky.callbacks.best_effort import BestEffortCallback
bec = BestEffortCallback()

# Send all metadata/data captured to the BestEffortCallback.
RE.subscribe(bec)
```

## Data broker

```python
from databroker import Broker
db = Broker.named('temp')

# Insert all metadata/data captured into db.
RE.subscribe(db.insert)
```

```python
from bluesky.utils import ProgressBarManager
RE.waiting_hook = ProgressBarManager()
```

```python
from ophyd.sim import det1, det2  # two simulated detectors
```
