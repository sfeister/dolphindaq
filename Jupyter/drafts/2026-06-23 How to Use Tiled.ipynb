# Tiled Example

Following a Bluesky example on saving data from a run using Tiled

## References
1. [Bluesky Tiled Tutorial](https://blueskyproject.io/bluesky/main/tiled-writer.html) (scroll to the bottom to get to the code for this example)

## Imports

```python
from bluesky import RunEngine
import bluesky.plans as bp
from bluesky.callbacks.tiled_writer import TiledWriter
from tiled.server import SimpleTiledServer
from tiled.client import from_uri
from ophyd.sim import det
from ophyd.sim import hw
import os
```

```python
# Initialize the Tiled server and client
save_path = os.path.join(".", "outputs")
tiled_server = SimpleTiledServer(save_path, readable_storage=[save_path])
tiled_client = from_uri(tiled_server.uri)
```

```python
# Initialize the RunEngine and subscribe TiledWriter
RE = RunEngine()
tw = TiledWriter(tiled_client)
RE.subscribe(tw)
```

```python
# Run an experiment collecting internal data
uid, = RE(bp.count([det], 3))
data = tiled_client[uid]['primary/det'].read()
```

```python
data
```

```python
tiled_client[uid]['primary/det'].export("output_data.json")
```

```python
# Run an experiment collecting external data
uid, = RE(bp.count([hw(save_path=save_path).img], 2))
data = tiled_client[uid]['primary/img'].read()
```
