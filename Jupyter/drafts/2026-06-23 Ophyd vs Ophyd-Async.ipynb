## Ophyd Hello, World

```python
from ophyd.signal import EpicsSignal, EpicsSignalRO

trig_count = EpicsSignalRO(
    "PULSEGEN:trigger:count",
    name="trig_count"
)

trig_count.wait_for_connection()
trig_count.get()
```

## Ophyd-Async Hello, World

```python
from ophyd_async.epics.core import epics_signal_r

trig_count = epics_signal_r(
    int,
    "ca://PULSEGEN:trigger:count",
    name="trig_count",
)

await trig_count.connect()
await trig_count.get_value()
```
